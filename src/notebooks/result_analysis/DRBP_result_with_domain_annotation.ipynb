```python
# 파일 리스트 수집
rna_files = glob.glob('/Users/khj/Desktop/Project_ongoing/DRBP/data/DRBP_result_analysis/Binding_info/IPR_binding_info/RNA-binding_domain_IPR/*.tsv')  # 디렉토리 경로를 원하는 경로로 수정

# 각 파일을 읽어서 리스트에 담기
rna_df_list = [pd.read_csv(f, sep='\t') for f in rna_files]

# 하나의 DataFrame으로 합치기
rna_combined_df = pd.concat(rna_df_list, ignore_index=True)

# (선택) index 리셋
rna_combined_df.reset_index(drop=True, inplace=True)
RBP_IPR_list = list(rna_combined_df['Accession'])
len(RBP_IPR_list)
```

```python
# 파일 리스트 수집
dna_GO_files = glob.glob('/Users/khj/Desktop/Project_ongoing/DRBP/data/DRBP_result_analysis/Binding_info/GO_binding_info/DNA_binding_GO/*.json')  # 디렉토리 경로를 원하는 경로로 수정

# 각 파일을 읽어서 리스트에 담기
dna_GO_df_list = [pd.read_json(f) for f in dna_GO_files]
dna_merged_df = pd.concat(dna_GO_df_list, ignore_index=True)
dna_merged_df['id'] = dna_merged_df['results'].apply(lambda x: x.get('id') if isinstance(x, dict) else None)
dna_merged_df['descendants'] = dna_merged_df['results'].apply(lambda x: x.get('descendants') if isinstance(x, dict) else None)
dna_expanded_df = dna_merged_df.explode('descendants', ignore_index=True)
DNA_Binding_GO_list = list(dna_expanded_df['id'])+list(dna_expanded_df['descendants'])
DNA_Binding_GO_list = list(set(DNA_Binding_GO_list))
DNA_Binding_GO_list
```

```python
# 파일 리스트 수집
rna_GO_files = glob.glob('/Users/khj/Desktop/Project_ongoing/DRBP/data/DRBP_result_analysis/Binding_info/GO_binding_info/RNA_binding_GO/*.json')  # 디렉토리 경로를 원하는 경로로 수정

# 각 파일을 읽어서 리스트에 담기
rna_GO_df_list = [pd.read_json(f) for f in rna_GO_files]
rna_merged_df = pd.concat(rna_GO_df_list, ignore_index=True)
rna_merged_df['id'] = rna_merged_df['results'].apply(lambda x: x.get('id') if isinstance(x, dict) else None)
rna_merged_df['descendants'] = rna_merged_df['results'].apply(lambda x: x.get('descendants') if isinstance(x, dict) else None)
rna_expanded_df = rna_merged_df.explode('descendants', ignore_index=True)
RNA_Binding_GO_list = list(rna_expanded_df['id'])+list(rna_expanded_df['descendants'])
RNA_Binding_GO_list = list(set(RNA_Binding_GO_list))
RNA_Binding_GO_list
```

```python
DNA_IPR_GO_binding = interproscan[
    #(interproscan[11].isin(DBP_IPR_list)) |
    (interproscan[13].astype(str).apply(lambda x: any(go in x for go in DNA_Binding_GO_list)))
]
DNA_IPR_GO_binding
```

```python
DNA_IPR_GO_binding.drop_duplicates(subset=0)
```

```python
RNA_IPR_GO_binding = interproscan[
    #(interproscan[11].isin(RBP_IPR_list)) |
    (interproscan[13].astype(str).apply(lambda x: any(go in x for go in RNA_Binding_GO_list)))
]
RNA_IPR_GO_binding
```

```python
RNA_IPR_GO_binding.drop_duplicates(subset=0)
```

```python
D_B_ENSP = list(DNA_IPR_GO_binding.drop_duplicates(subset=0)[0])
R_B_ENSP = list(RNA_IPR_GO_binding.drop_duplicates(subset=0)[0])

print(f'DNA binding ENSP : {len(D_B_ENSP)}\nRNA binding ENSP : {len(R_B_ENSP)}')
```
