```python
import pandas as pd
import numpy as np
import glob
import os
import re
```

```python
# VAT score 파일들이 들어있는 디렉토리 경로
folder_path = '/Users/khj/Desktop/Project_ongoing/DRBP/data/DRBP_result_analysis/analysis_raw_data/VAT score/VATP_raw/DBP_VATP_result_rand10_251030'  # ⚠️ 실제 경로로 변경하세요

# CSV 파일 목록 불러오기
file_list = sorted(glob.glob(os.path.join(folder_path, '*.csv')))
print(f"총 {len(file_list)}개의 VAT score 파일 발견됨.")

all_data = []

for file in file_list:
    df = pd.read_csv(file)
    
    # 파일명에서 ENSP ID 추출
    # 예: '>ENSP00000000233_VATP_frontPAD40_mask0.csv' → 'ENSP00000000233'
    base_name = os.path.basename(file)
    match = re.search(r'ENSP\d+', base_name)
    if match:
        ensp_id = match.group(0)
    else:
        print(f"⚠️ ENSP ID를 찾을 수 없음: {file}")
        continue

    # 열 이름을 ENSP ID로 통일
    df.columns = [f"{col}_{ensp_id}" for col in df.columns]
    all_data.append(df.reset_index(drop=True))

# 열 방향(axis=1)으로 합치기
merged_df = pd.concat(all_data, axis=1)

# 결과 저장
out_path = os.path.join('/Users/khj/Desktop/Project_ongoing/DRBP/data/DRBP_result_analysis/analysis_raw_data/VAT score/VATP_raw/', 'DBP_model_VATP_rand10_251030.csv')
merged_df.to_csv(out_path, index=False)

print(f"✅ 병합 완료: {out_path}")
```

```python
# 파일 불러오기
dbp_VATP = pd.read_csv('/Users/khj/Desktop/Project_ongoing/DRBP/data/DRBP_result_analysis/analysis_raw_data/VAT score/VATP_raw/DBP_model_VATP_251030.csv')

# 실제 서열 구간만 남기기 (앞 40 pad + CLS 제거)
dbp_filtered_VATP_data = dbp_VATP.iloc[41:1023, 1::2].copy()
dbp_filtered_VATP_data.columns = dbp_filtered_VATP_data.columns.str.replace('VATP_', '', regex=False)

# Min–Max scaling per protein
dbp_normalized_df = dbp_filtered_VATP_data.copy()

for col in dbp_normalized_df.columns:
    min_val = dbp_normalized_df[col].min()
    max_val = dbp_normalized_df[col].max()
    if pd.isna(min_val) or pd.isna(max_val) or max_val == min_val:
        # 예외 처리: 값이 모두 같거나 NaN일 경우 0으로 설정
        dbp_normalized_df[col] = 0
    else:
        dbp_normalized_df[col] = (dbp_normalized_df[col] - min_val) / (max_val - min_val)

dbp_normalized_df.to_csv('/Users/khj/Desktop/Project_ongoing/DRBP/data/DRBP_result_analysis/analysis_raw_data/VAT score/VATP_raw/DBP_model_VATP_251030_min_max_normalized.csv', index=False)
```

```python
# CSV 파일들이 있는 디렉토리 경로
directory = '/Users/khj/Desktop/Project_ongoing/DRBP/data/DRBP_result_analysis/analysis_raw_data/VAT score/VATP_raw/DNA model/NDBP'

# 디렉토리 내 모든 csv 파일 목록 불러오기
csv_files = [f for f in os.listdir(directory) if f.endswith(".csv")]

# 각 CSV 파일을 데이터프레임으로 읽어서 리스트에 저장
dfs = [pd.read_csv(os.path.join(directory, f)) for f in csv_files]

# 열 기준(axis=1)으로 병합
merged_df = pd.concat(dfs, axis=1)

pad_VAT = merged_df.iloc[:, 1::2].copy()
pad_VAT.columns = pad_VAT.columns.str.replace('VATP_', '', regex=False)

pad_VAT_clean = pad_VAT.copy()
pad_VAT_clean = pad_VAT_clean.replace(0, np.nan)

for col in pad_VAT_clean.columns:
    # 해당 열의 마지막 유효값 위치 찾기
    last_valid = pad_VAT_clean[col].last_valid_index()
    # 그 위치의 값 제거 (NaN으로 설정)
    if last_valid is not None:
        pad_VAT_clean.loc[last_valid, col] = np.nan

pad_VAT_clean.to_csv('/Users/khj/Desktop/Project_ongoing/DRBP/data/DRBP_result_analysis/analysis_raw_data/VAT score/VATP_raw/NDBP_original_VATP.csv',index=None)
```
